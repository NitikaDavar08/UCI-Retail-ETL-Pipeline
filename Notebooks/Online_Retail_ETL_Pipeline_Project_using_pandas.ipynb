### **Now we're going to move from checking data quality to actually cleaning and transforming the data. In a real business, we don't delete every unusual record just because it looks incorrect.we'll preserve the original data and create a separate cleaned dataset. This is how we'll avoid accidentally destroying useful business information.**

Step 1: Create a working copy. Why are we doing this?
Our original df contains the raw business data. We should never directly modify our only copy.

We'll create another DataFrame called df_clean.

```python
# Create a separate copy of the original dataset
df_clean = df.copy()

# Display the number of rows and columns in the new dataset
print("Working dataset shape:", df_clean.shape)

# df.copy()	Creates a separate copy of the DataFrame
# df_clean	Stores our working dataset
# shape	Returns the number of rows and columns
# print()	Displays the result
```

Previously we had counted duplicates. But we didn't remove them because we had to preserve the original data.

But now we will be cleaning and transforming data. Therefore we'll remove exact duplicate rows.

```python
# Count duplicate records before removing them
duplicates_before = df_clean.duplicated().sum()

# Remove exact duplicate rows and keep the first occurrence
df_clean = df_clean.drop_duplicates()

# Count duplicate records after cleaning
duplicates_after = df_clean.duplicated().sum()

# Display the cleaning results
print("Duplicates before cleaning:", duplicates_before)
print("Duplicates after cleaning:", duplicates_after)

# drop_duplicates() removes the repeated row and keeps one copy.
# Important: We are removing only completely identical rows, not every row with
#  the same InvoiceNo. One invoice can legitimately contain several products.
```

invoices beginning with C represent cancellations. This is part of the UCI dataset's official description.

Should we delete them?
Not immediately. A business may want to analyse cancelled orders separately.

So we'll create two datasets:
- df_cancelled: Cancelled transactions.
- df_sales: Transactions that are not marked as cancelled.

```python
# Identify invoices that start with the letter C
cancellation_mask = (
    df_clean["InvoiceNo"].astype(str).str.startswith("C")
)

# Store cancelled transactions separately
df_cancelled = df_clean[cancellation_mask].copy()

# Keep transactions that are not marked as cancelled
df_sales = df_clean[~cancellation_mask].copy()

# Display the number of records in both datasets
print("Cancelled records:", len(df_cancelled))
print("Non-cancelled records:", len(df_sales))

# - df_clean["InvoiceNo"]: Selects the invoice number column.
# - astype(str): Converts invoice numbers to text.
# - str.startswith("C"): Checks whether an invoice starts with C.
# - cancellation_mask: Stores True or False for each row.
# - df_clean[condition]: Selects rows matching the condition.
# - ~: Reverses the condition. True becomes False and vice versa.
# - .copy(): Creates a separate copy of the selected records.
# - len(): Counts the records.

# Why are we creating two datasets?
# Because cancelled transactions and successful sales have different business meanings. We want to preserve both for separate analysis.
```

Separate returns and unusual records
Remember our negative quantity records?

A negative quantity can represent returned products. We should not automatically treat it as an error.

Let's separate them.

```python
# Identify transactions with negative quantities
returns_mask = df_sales["Quantity"] < 0

# Store negative quantity records separately
df_returns = df_sales[returns_mask].copy()

# Keep positive quantity records for the sales dataset
df_sales = df_sales[df_sales["Quantity"] > 0].copy()

# Display the number of sales and return records
print("Return records:", len(df_returns))
print("Positive quantity sales records:", len(df_sales))

# If we delete the return record, our business analysis becomes incomplete.
# That's why we retain returns separately.
# We are also excluding zero-quantity records from the positive-sales dataset.
```

identify invalid prices
A transaction with a zero or negative unit price cannot be treated as an ordinary priced sale.

We'll separate these records rather than permanently delete them.

Why are we doing this?

Because we want to calculate sales revenue using:
Revenue = Quantity * Unit Price
For our normal sales dataset, both values should be positive. We are preserving the unusual records for further investigation.

```python
# Identify records with zero or negative unit prices
invalid_price_mask = df_sales["UnitPrice"] <= 0

# Store invalid price records separately
df_invalid_prices = df_sales[invalid_price_mask].copy()

# Keep records with positive unit prices
df_sales = df_sales[df_sales["UnitPrice"] > 0].copy()

# Display the results
print("Invalid price records:", len(df_invalid_prices))
print("Valid price records:", len(df_sales))
```

*# Transform the sales data We're going to create new information from the existing business records*.

Convert InvoiceDate into a proper date format. Why?
Our original CSV may store the invoice date as text or as an object. We want Python to understand it as an actual date and time.

```python
# Convert the InvoiceDate column into datetime format
df_sales["InvoiceDate"] = pd.to_datetime(
    df_sales["InvoiceDate"],
    errors="coerce"
)

# Display the data types after conversion
print(df_sales.dtypes)

# pd.to_datetime(): Converts values into date and time format.
# errors="coerce": Converts invalid date values into missing values (NaT).
# dtypes: Displays the data type of every column.
```

Calculate total sales revenue
Now imagine your manager asks:

"How much revenue did we generate from each transaction line?"

We already have:
- Quantity
- UnitPrice
So we can calculate revenue.

```python
# Calculate the total revenue for each transaction line
df_sales["TotalSales"] = (
    df_sales["Quantity"] * df_sales["UnitPrice"]
)

# Display the first five records
print(df_sales.head())

# Now our dataset has a new column called TotalSales.
```

Step 1: Check missing values again

```python
# Count missing values in each column of the cleaned sales dataset
missing_values = df_sales.isnull().sum()

# Display the missing value count for each column
print(missing_values)

# Command logic:
# - df_sales: Our cleaned sales DataFrame.
# - isnull(): Identifies missing values.
# - sum(): Counts missing values column-wise.
# - missing_values: Stores the result.
```

Step 2: Check missing Customer IDs
Our retail dataset contains a CustomerID column.

Imagine a customer purchases a product, but their customer ID is missing.

Can we identify which customer made that purchase?
No.
But does that mean the sale itself is invalid?

Not necessarily! We should preserve the transaction for sales analysis while recognising that customer-level analysis may not be possible for that record.

```python
# Count records where CustomerID is missing
missing_customer_count = df_sales["CustomerID"].isnull().sum()

# Display the number of records with missing customer IDs
print("Missing Customer IDs:", missing_customer_count)
```

Step 3: Remove records with missing essential information
For our sales analysis, we need:
- InvoiceNo
- StockCode
- Quantity
- UnitPrice
- InvoiceDate
If any of these are missing, we cannot reliably calculate or identify a transaction line.

```python
# Define the columns that must contain valid information
required_columns = [
    "InvoiceNo",
    "StockCode",
    "Quantity",
    "UnitPrice",
    "InvoiceDate"
]

# Remove records with missing values in these essential columns
df_sales = df_sales.dropna(subset=required_columns).copy()

# Display the updated dataset dimensions
print("Dataset shape after validation:", df_sales.shape)

# - required_columns: A list of essential column names.
# - dropna(): Removes rows containing missing values in the specified columns.
# - subset: Tells Python which columns to check.
# - copy(): Creates a separate DataFrame.
# Notice that we are not including CustomerID in the required columns.
```

- TotalSales should be numeric.

```python
# Display the data types of all columns
print(df_sales.dtypes)
```

Step 5: Check whether our cleaning was successful

Step 6: Calculate total business revenue
Now let's calculate the total revenue generated by the remaining valid sales records

```python
# Calculate the total revenue from all valid sales records
total_revenue = df_sales["TotalSales"].sum()

# Display the total revenue
print("Total Business Revenue:", total_revenue)

# - df_sales["TotalSales"] selects the revenue column.
# - .sum() adds the revenue values.
# - total_revenue stores the final amount.
# This gives us an important business metric.
```

Step 7: Analyse revenue by country
Imagine the company's manager asks:

"Which country generates the highest sales revenue?"
We can answer using Pandas.

```python
# Group all sales records according to country
# Calculate the total revenue generated by each country
country_revenue = df_sales.groupby("Country")["TotalSales"].sum()

# Sort the countries from highest revenue to lowest revenue
country_revenue = country_revenue.sort_values(ascending=False)

# Display the top 10 countries
print(country_revenue.head(10))

# groupby("Country")	Groups records by country
# ["TotalSales"]	Selects the revenue column
# sum()	Calculates total revenue per country
# sort_values()	Arranges the results
# ascending=False	Sorts from highest to lowest
# head(10)	Displays the top 10 results
```

Step 8: Prepare the final cleaned dataset
Now we need to create a final dataset that will eventually be loaded into Oracle.

Before that, let's check whether our revenue column is present and correctly calculated.

```python
# Recalculate revenue to ensure it matches quantity multiplied by unit price
df_sales["TotalSales"] = (
    df_sales["Quantity"] * df_sales["UnitPrice"]
)

# Display the first five records
print(df_sales.head())

# Why recalculate?
# Because we have removed some records and performed several cleaning operations. Recalculating ensures the revenue column follows our business formula.
```

Step 9: Save the cleaned dataset as a new CSV file
This is important!

Our original CSV is raw data. We now want to create a separate file containing our transformed data.

```python
import pandas as pd
import os

# If df_sales is not in memory, reconstruct it from the uploaded file
if 'df_sales' not in globals():
    print("df_sales not found in memory. Reconstructing from the uploaded file...")
    # Locate the CSV file in the current directory, ignoring the cleaned output filename
    csv_files = [f for f in os.listdir('.') if f.endswith('.csv') and f != 'cleaned_online_retail.csv']

    if not csv_files:
        print("No raw CSV files found in the current directory. Let's look at the active files:", os.listdir('.'))
        raise FileNotFoundError("Please make sure your raw 'Online Retail' CSV file is uploaded to Colab. Use the upload cell (Step 1) to upload it first.")

    # Use the first available raw CSV file
    raw_filename = csv_files[0]
    print(f"Reading raw dataset from: {raw_filename}")
    df = pd.read_csv(raw_filename)

    # 1. Remove duplicates
    df_clean = df.drop_duplicates()

    # 2. Separate cancellations
    cancellation_mask = df_clean["InvoiceNo"].astype(str).str.startswith("C")
    df_sales = df_clean[~cancellation_mask].copy()

    # 3. Filter valid quantities and unit prices
    df_sales = df_sales[(df_sales["Quantity"] > 0) & (df_sales["UnitPrice"] > 0)].copy()

    # 4. Convert date and drop rows missing essential values
    df_sales["InvoiceDate"] = pd.to_datetime(df_sales["InvoiceDate"], errors="coerce")
    required_columns = ["InvoiceNo", "StockCode", "Quantity", "UnitPrice", "InvoiceDate"]
    df_sales = df_sales.dropna(subset=required_columns).copy()

    # 5. Calculate total sales revenue
    df_sales["TotalSales"] = df_sales["Quantity"] * df_sales["UnitPrice"]

# Save the cleaned DataFrame as a new CSV file
df_sales.to_csv("cleaned_online_retail.csv", index=False)
print("Cleaned dataset saved successfully as 'cleaned_online_retail.csv'!")
```

```python
import pandas as pd
from google.colab import files
import os

# 1. Ask for upload if the raw dataset is not in the directory
if not any(f.endswith('.csv') and f != 'cleaned_online_retail.csv' for f in os.listdir('.')):
    print("Please upload your raw Online Retail CSV file below:")
    files.upload()

# 2. Find the uploaded file name
filename = [f for f in os.listdir('.') if f.endswith('.csv') and f != 'cleaned_online_retail.csv'][0]

# 3. Quick Clean & Save
df = pd.read_csv(filename)
df_clean = df.drop_duplicates()
df_sales = df_clean[~df_clean["InvoiceNo"].astype(str).str.startswith("C")].copy()
df_sales = df_sales[(df_sales["Quantity"] > 0) & (df_sales["UnitPrice"] > 0)].copy()

df_sales["InvoiceDate"] = pd.to_datetime(df_sales["InvoiceDate"], errors="coerce")
df_sales = df_sales.dropna(subset=["InvoiceNo", "StockCode", "Quantity", "UnitPrice", "InvoiceDate"]).copy()
df_sales["TotalSales"] = df_sales["Quantity"] * df_sales["UnitPrice"]

# 4. Save and Download
df_sales.to_csv("cleaned_online_retail.csv", index=False)
print("Done! Downloading your cleaned file...")
files.download("cleaned_online_retail.csv")
```
